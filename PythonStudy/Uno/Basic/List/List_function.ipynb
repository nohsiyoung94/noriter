```python
a = [1,2,3,4,5]
del a[2:]
print(a)
```

```python
#append
a = [1,2,3]
a.append(4)
print(a)
```

```python
#sort
a = [1,4,2,3]
a.sort()
print(a)
```

```python
a= ['a','b','d','c']
print(a)
a.sort()
print(a)
```

```python
a = [1,2,3]
print(a.index(3))
print(a.index(2))
print(a.index(1))
print(a.index(0))
print(a)
```

```python
a = [1,2,3]
a.insert(3,5)
print(a)
```

```python
#remove
a = [1,2,3,1,2,3]
print(a)
a.remove(3)
print(a)
a.remove(3)
print(a)
```

```python
#pop
a = [1,2,3,4,5]
a.pop()
b= 2
print(a)
a.pop(b)
print(a)
```

```python
#count
a = [1,2,3,4,5,3]
print(a.count(3))
print(a)
```

```python
#extend
a = [1,2,3]
a.extend([4,5])
print(a)
b = [6,7]
a.extend(b)
print(a)
```
